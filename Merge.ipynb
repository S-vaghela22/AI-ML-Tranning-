```python
import pandas as pd

# Dictionary for DataFrame 1
student_data = {
    "Student_ID": [101, 102, 103],
    "Name": ["Amit", "Priya", "Rahul"],
    "Age": [21, 22, 20]
}

# Dictionary for DataFrame 2
marks_data = {
    "Student_ID": [101, 102, 103],
    "Maths": [85, 90, 78],
    "Science": [88, 92, 81]
}

# Create DataFrames
df1 = pd.DataFrame(student_data)
df2 = pd.DataFrame(marks_data)

# Display DataFrames
print("DataFrame 1:")
print(df1)

print("\nDataFrame 2:")
print(df2)
```

```python
merge_df = pd.merge(df1, df2, on="Student_ID")
print("\nMerged DataFrame:")
merge_df
```

```python
merge_df = pd.merge(df1, df2, on="Student_ID",how='inner')
print("\nMerged DataFrame:")
merge_df
```

```python
import pandas as pd

# Dictionary for DataFrame 1
student_data = {
    "Student_ID": [101, 102, 103],
    "Name": ["Amit", "Priya", "Rahul"],
    "Age": [21, 22, 20]
}

# Dictionary for DataFrame 2
marks_data = {
    "Student_ID": [101, 102, 103],
    "Maths": [85, 90, 78],
    "Science": [88, 92, 81]
}

# Create DataFrames
df1 = pd.DataFrame(student_data)
df2 = pd.DataFrame(marks_data)

# Inner Join
result = pd.merge(df1, df2, on="Student_ID", how="inner")

print(result)
```

```python
merge_df = pd.merge(df1, df2, on="Student_ID",how='outer')
print("\nMerged DataFrame:")
merge_df
```

```python
df1 = pd.DataFrame({
    'User_ID':[101,102,103],
    'Name':['Alice','Bob','Charlie']
}, index=[11,22,33])
print(df1)

df2 = pd.DataFrame({
    'User_ID':[101,102,106],
    'City':['New York','London','Paris']
}, index=[100,200,300])
print(df2)
```

```python
index_access = pd.merge(df1,df2,left_index=True,right_index=True)
index_access
```
