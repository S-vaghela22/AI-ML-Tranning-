```python
tips_df = sns.load_dataset('tips')
tips_df
```

```python
sns.lineplot(x='total_bill',y="tip",data=tips_df)#space also include
```

```python
sns.lineplot(x='size',
             y='total_bill',
             data=tips_df,
             hue = 'sex')
```

```python
sns.lineplot(x='size',
             y='total_bill',
             data=tips_df,
             hue = 'sex' ,
             style='sex')
```

```python
sns.lineplot(x='size',
             y='total_bill',
             data=tips_df,
             hue = 'sex' ,
             style='sex',
             palette="cool")
```

```python
sns.lineplot(x='size',
             y='total_bill',
             data=tips_df,
             hue = 'sex' ,
             style='sex',
             palette="hot")
```

```python
sns.lineplot(x='size',y='total_bill',data=tips_df,hue = 'sex' ,style='sex',palette="hot",dashes=False)
```

```python
sns.lineplot(x='size',
             y='total_bill',
             data=tips_df,
             hue = 'sex' ,
             style='sex',
             palette="hot",
             dashes=False,
             markers=True,
             legend =  False)
```

```python
sns.lineplot(x='size',
             y='total_bill',
             data=tips_df,
             hue = 'sex' ,
             style='sex',
             palette="hot",
             dashes=False,
             markers=True,
             legend =  'brief')
```

```python
plt.figure(figsize=(10,9))
sns.set(style='darkgrid')
sns.lineplot(x='size',
             y='total_bill',
             data=tips_df,
             hue = 'day' ,
             style='day',
             palette="hot",
             dashes=False,
             markers=['o','<','>','^'],
             legend =  'brief',)

plt.title('Line Plot',fontsize =19)
plt.xlabel('Size',fontsize=15)
plt.ylabel('Total Bill',fontsize=15)
plt.show()
```
