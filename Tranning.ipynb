```python
#WAP to print Compound Interest?

p = float(input("Enter Principal: "))
r = float(input("Enter Rate of Interest: "))
t = float(input("Enter Time: "))

amount = p * (1 + r/100) ** t
ci = amount - p

print("Compound Interest is:", ci)
```
