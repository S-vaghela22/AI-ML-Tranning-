```python
classes = ['Python','R','AI','ML','DS']
class1_student = [45,17,33,30,29]
class2_students = [4,5,20,20,10]
class3_students =  [3,5,30,15,15]

explode = [0,0,0.2,0.1,0.3]
colors = ['c','b','r','y','g']
textprops = {'fontsize':60}

plt.pie(class1_student , labels= classes,explode=explode,colors=colors,autopct='%0.if%%',shadow=True,radius=1.4,startangle=280)
plt.show()

```

```python
classes = ['Python','R','AI','ML','DS']
class1_student = [45,17,33,30,29]
class2_students = [4,5,20,20,10]
class3_students =  [3,5,30,15,15]

explode = [0,0,0.2,0.1,0.3]
colors = ['c','b','r','y','g']
textprops = {'fontsize':6}

wedgeprops = {'linewidth':4,'width':1,'edgecolor':'k'}

plt.pie(
    class1_student ,
    labels= classes,
    explode=explode,
    colors=colors,
    autopct='%0.if%%',
    pctdistance=0.6,
    shadow=True,
    labeldistance=1.4,
    counterclock=True,
    wedgeprops=wedgeprops,
    textprops=textprops,
    center = (2,3),
    frame=True,
    radius=1.4,
    rotatelabels=True,
    startangle=280,
)
plt.show()
```
