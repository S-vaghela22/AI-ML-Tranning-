```python
df = pd.read_csv("googleplaystore.csv", nrows=1000)
df
```

```python
x =df['Rating']
y = df['Reviews']
```

```python
import matplotlib.pyplot as plt
plt.scatter(x,y)
plt.show()
```

```python
plt.scatter(x,y)

plt.xlabel("Rating")
plt.ylabel("Reviews")
plt.title("Google play store Apps & Scatter Plot")
plt.show()
```

```python
plt.scatter(x,y, color = 'r')

plt.xlabel("Rating")
plt.ylabel("Reviews")
plt.title("Google play store Apps & Scatter Plot")
plt.show()
```

```python
plt.figure(figsize=(7,5))
plt.scatter(x,y, color = 'r',marker='*')

plt.xlabel("Rating")
plt.ylabel("Reviews")
plt.title("Google play store Apps & Scatter Plot")
plt.show()
```

```python
plt.figure(figsize=(7,5))
plt.scatter(x,y, color = 'r',marker='*',s= 100,alpha=None,linewidths=5,edgecolors='b')

plt.xlabel("Rating")
plt.ylabel("Reviews")
plt.title("Google play store Apps & Scatter Plot")
plt.show()
```

```python
plt.figure(figsize=(16,7))
plt.scatter(x,y, color = 'r',marker='*',s= 100,alpha=None,linewidths=5,edgecolors='b')
plt.scatter(x,df['Installs'],c='y',marker='o',s=80,alpha=0.5,linewidths=5,edgecolors='c')

plt.xlabel("Rating")
plt.ylabel("Installs")
plt.title("Google play store Apps & Scatter Plot")
plt.show()
```
