```python
m1_students_age = np.random.randint(18,50,(100))
py_students_age = np.random.randint(14,50,(100))

bins = [15,20,25,30,35,40,45]
plt.hist(m1_students_age,bins)
plt.title("M1 Students Age")
plt.xlabel("Age")
plt.ylabel("Number of Students")
plt.show()
```

```python
plt.hist(m1_students_age,bins,rwidth=0.90)
plt.title("M1 Students Age")
plt.xlabel("Age")
plt.ylabel("Number of Students")
plt.show()
```

```python
plt.hist(m1_students_age,bins,rwidth=0.90,histtype='step')
plt.title("M1 Students Age")
plt.xlabel("Age")
plt.ylabel("Number of Students")
plt.show()
```

```python
plt.hist(m1_students_age,bins,rwidth=0.90,histtype='barstacked')
plt.title("M1 Students Age")
plt.xlabel("Age")
plt.ylabel("Number of Students")
plt.show()
```

```python
plt.hist(m1_students_age,bins,rwidth=0.90,histtype='barstacked',align='left')
plt.title("M1 Students Age")
plt.xlabel("Age")
plt.ylabel("Number of Students")
plt.show()
```

```python
plt.hist(m1_students_age,bins,rwidth=0.90,histtype='barstacked',align='right')
plt.title("M1 Students Age")
plt.xlabel("Age")
plt.ylabel("Number of Students")
plt.show()
```

```python
plt.hist(m1_students_age,bins,rwidth=0.90,histtype='barstacked',orientation='horizontal')
plt.title("M1 Students Age")
plt.xlabel("Age")
plt.ylabel("Number of Students")
plt.show()
```

```python
plt.hist(m1_students_age,bins,rwidth=0.90,histtype='barstacked',color='y')
plt.title("M1 Students Age")
plt.xlabel("Age")
plt.ylabel("Number of Students")
plt.show()
```

```python
plt.hist(m1_students_age,bins,rwidth=0.90,histtype='barstacked',color='y',label='ML-student')
plt.title("M1 Students Age")
plt.xlabel("Age")
plt.ylabel("Number of Students")
plt.legend()
plt.show()
```

```python
bins =[15,20,25,30,35,40,45]

plt.hist(m1_students_age,bins,rwidth=0.90,histtype='bar',orientation='horizontal',color='y',label='ML-student')
plt.hist(py_students_age,bins,rwidth=0.90,histtype='bar',orientation='horizontal',color='orange',label='py-student')

plt.title("M1 Students Age")
plt.xlabel("Age")
plt.ylabel("Number of Students")
plt.legend()
plt.show()
```

```python
bins =[15,20,25,30,35,40,45]

plt.hist(m1_students_age,bins,rwidth=0.90,histtype='bar',orientation='horizontal',color='y',label='ML-student')
plt.hist(py_students_age,bins,rwidth=0.90,histtype='bar',orientation='vertical',color='orange',label='py-student')

plt.title("M1 Students Age")
plt.xlabel("Age")
plt.ylabel("Number of Students")
plt.legend()
plt.show()
```

```python
bins =[15,20,25,30,35,40,45]

plt.hist(m1_students_age,bins,rwidth=0.90,histtype='bar',color='y',label='ML-student')
plt.hist(py_students_age,bins,rwidth=0.90,histtype='bar',color='orange',label='py-student')

plt.title("M1 Students Age")
plt.xlabel("Age")
plt.ylabel("Number of Students")
plt.legend()
plt.show()
```

```python
bins =[15,20,25,30,35,40,45]

plt.hist([m1_students_age, py_students_age], bins, rwidth=0.90, histtype='bar', color=['y','b'], label=['ML-student','py-student'])

plt.title("Students Age Distribution")
plt.xlabel("Number of Students")
plt.ylabel("Age")
plt.legend()
plt.show()
```

```python
bins = [15,20,25,30,35,40,45]

plt.hist([m1_students_age, py_students_age], bins=bins, rwidth = 0.8, histtype = 'bar',
        color = ['r','b'], label = ['ML Students','Python Students'])

plt.title('ML STUDENTS AGE')
plt.xlabel('AGE')
plt.ylabel('NUMBER OF STUDENTS AGE')
plt.legend()
plt.show()
```
